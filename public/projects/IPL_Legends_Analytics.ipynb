```python
# ==========================================
# IPL LEGENDS ANALYTICS PROJECT
# Virat Kohli vs Rohit Sharma vs MS Dhoni
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

# ==========================================
# LOAD DATASETS
# ==========================================

player = pd.read_csv(r"C:\DS\player_summary.csv")
season = pd.read_csv(r"C:\DS\season_wise_stats.csv")
captain = pd.read_csv(r"C:\DS\captaincy_stats.csv")

# ==========================================
# DISPLAY TABLES
# ==========================================

print("="*60)
print("PLAYER SUMMARY")
print("="*60)
display(player)

print("="*60)
print("SEASON WISE DATA")
print("="*60)
display(season)

print("="*60)
print("CAPTAINCY DATA")
print("="*60)
display(captain)

# ==========================================
# GRAPH 1 : TOTAL RUNS
# ==========================================

plt.figure(figsize=(8,5))

plt.bar(
    player["Player"],
    player["Total_Runs"]
)

plt.title("Total IPL Runs")
plt.xlabel("Player")
plt.ylabel("Runs")

plt.show()

# ==========================================
# GRAPH 2 : STRIKE RATE
# ==========================================

plt.figure(figsize=(8,5))

plt.bar(
    player["Player"],
    player["Average_StrikeRate"]
)

plt.title("Average Strike Rate")
plt.xlabel("Player")
plt.ylabel("Strike Rate")

plt.show()

# ==========================================
# GRAPH 3 : FIFTIES
# ==========================================

plt.figure(figsize=(8,5))

plt.bar(
    player["Player"],
    player["Fifties"]
)

plt.title("Total Fifties")

plt.show()

# ==========================================
# GRAPH 4 : HUNDREDS
# ==========================================

plt.figure(figsize=(8,5))

plt.bar(
    player["Player"],
    player["Hundreds"]
)

plt.title("Total Hundreds")

plt.show()

# ==========================================
# GRAPH 5 : FOURS
# ==========================================

plt.figure(figsize=(8,5))

plt.bar(
    player["Player"],
    player["Fours"]
)

plt.title("Total Fours")

plt.show()

# ==========================================
# GRAPH 6 : SIXES
# ==========================================

plt.figure(figsize=(8,5))

plt.bar(
    player["Player"],
    player["Sixes"]
)

plt.title("Total Sixes")

plt.show()

# ==========================================
# GRAPH 7 : MATCHES
# ==========================================

plt.figure(figsize=(8,5))

plt.bar(
    player["Player"],
    player["Matches"]
)

plt.title("Total Matches")

plt.show()

# ==========================================
# GRAPH 8 : RUNS TREND
# ==========================================

plt.figure(figsize=(12,6))

plt.plot(
    season["Season"],
    season["Virat_Runs"],
    marker='o',
    label="Virat Kohli"
)

plt.plot(
    season["Season"],
    season["Rohit_Runs"],
    marker='o',
    label="Rohit Sharma"
)

plt.plot(
    season["Season"],
    season["Dhoni_Runs"],
    marker='o',
    label="MS Dhoni"
)

plt.legend()

plt.title("Season Wise Runs")

plt.xlabel("Season")
plt.ylabel("Runs")

plt.show()

# ==========================================
# GRAPH 9 : STRIKE RATE TREND
# ==========================================

plt.figure(figsize=(12,6))

plt.plot(
    season["Season"],
    season["Virat_SR"],
    marker='o',
    label="Virat Kohli"
)

plt.plot(
    season["Season"],
    season["Rohit_SR"],
    marker='o',
    label="Rohit Sharma"
)

plt.plot(
    season["Season"],
    season["Dhoni_SR"],
    marker='o',
    label="MS Dhoni"
)

plt.legend()

plt.title("Season Wise Strike Rate")

plt.xlabel("Season")
plt.ylabel("Strike Rate")

plt.show()

# ==========================================
# GRAPH 10 : WIN PERCENTAGE
# ==========================================

plt.figure(figsize=(8,5))

plt.bar(
    captain["Player"],
    captain["Win_Percentage"]
)

plt.title("Captaincy Win Percentage")

plt.ylabel("Win %")

plt.show()

# ==========================================
# GRAPH 11 : IPL TITLES
# ==========================================

plt.figure(figsize=(8,5))

plt.bar(
    captain["Player"],
    captain["IPL_Titles"]
)

plt.title("IPL Titles")

plt.ylabel("Titles")

plt.show()

# ==========================================
# GRAPH 12 : PIE CHART OF RUNS
# ==========================================

plt.figure(figsize=(8,8))

plt.pie(
    player["Total_Runs"],
    labels=player["Player"],
    autopct="%1.1f%%"
)

plt.title("Runs Distribution")

plt.show()

# ==========================================
# GRAPH 13 : HEATMAP
# ==========================================

heatmap_data = season.set_index("Season")

plt.figure(figsize=(12,6))

sns.heatmap(
    heatmap_data.corr(),
    annot=True
)

plt.title("Correlation Heatmap")

plt.show()

# ==========================================
# GRAPH 14 : BOXPLOT
# ==========================================

runs_data = [
    season["Virat_Runs"],
    season["Rohit_Runs"],
    season["Dhoni_Runs"]
]

plt.figure(figsize=(8,5))

plt.boxplot(
    runs_data,
    labels=[
        "Virat",
        "Rohit",
        "Dhoni"
    ]
)

plt.title("Runs Distribution")

plt.show()

# ==========================================
# GRAPH 15 : GOAT SCORE
# ==========================================

player["GOAT_Score"] = (
    player["Total_Runs"] * 0.4 +
    player["Average_StrikeRate"] * 5 +
    player["Fifties"] * 20 +
    player["Hundreds"] * 50
)

print("="*60)
print("GOAT TABLE")
print("="*60)

display(
    player[
        [
            "Player",
            "GOAT_Score"
        ]
    ]
)

plt.figure(figsize=(8,5))

plt.bar(
    player["Player"],
    player["GOAT_Score"]
)

plt.title("IPL GOAT Score")

plt.show()

# ==========================================
# FINAL RANKING
# ==========================================

ranking = player.sort_values(
    by="GOAT_Score",
    ascending=False
)

print("="*60)
print("FINAL GOAT RANKING")
print("="*60)

display(
    ranking[
        [
            "Player",
            "GOAT_Score"
        ]
    ]
)

print("\nPROJECT COMPLETED SUCCESSFULLY")
```
